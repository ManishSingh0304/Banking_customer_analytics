### Numerical Analysis

```python
df.columns
```

```python
numerical_cols = ['estimated_income','superannuation_savings','credit_card_balance','bank_loans','bank_deposits','checking_accounts','saving_accounts','foreign_currency_account','business_lending']
```

### Univariate Analysis and Visualization 

```python
plt.figure(figsize=(20,15))
for i,col in enumerate(numerical_cols):
    plt.subplot(4,3,i+1)
    sns.histplot(df[col],kde=True)
    plt.title(col)
plt.show()     
```

### Heatmaps

```python
correlation_matrix = df[numerical_cols].corr()
plt.figure(figsize=(12,12))
sns.heatmap(correlation_matrix,annot=True, cmap= 'crest',fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```

## Insights of EDA :-
#### Strong Relationship Between Bank Deposits and Checking Accounts
1. Correlation = 0.84
2. Customers with higher bank deposits tend to maintain larger balances in their checking accounts.
3. This is the strongest relationship in the dataset.
4. Indicates that checking accounts are a major channel for holding deposited funds.

#### Strong Relationship Between Bank Deposits and Saving Accounts
1. Correlation = 0.75
2. Customers who keep larger deposits are also more likely to maintain higher savings balances.
3. Suggests that deposit-heavy customers prefer diversified banking products.

#### Moderate Relationship Between Saving and Checking Accounts
1. Correlation = 0.46
2. Customers often use both savings and checking accounts together.
3. Cross-selling opportunities may exist between these products.

### Lending and Credit Insights

#### Business Lending Shows Moderate Correlation with Bank Deposits
1. Correlation = 0.44
2. Customers with larger deposits are somewhat more likely to use business lending services.
3. Deposit-rich customers may represent valuable lending prospects.

#### Business Lending and Bank Loans
1. Correlation = 0.42
2. Customers already utilizing loans tend to engage more with business lending products.
3. Existing borrowers may be good targets for additional lending offerings.

#### Credit Card Balance and Bank Deposits
1. Correlation = 0.38
2. Customers with larger deposit balances often maintain higher credit card balances.
3. Indicates greater overall financial activity.

### Income Insights

#### Estimated Income Has Weak-to-Moderate Correlations
###### Highest correlations:
1. Income ↔ Superannuation Savings (0.37)
2. Income ↔ Business Lending (0.33)
3. Income ↔ Bank Loans (0.33)

### Foreign Currency Account Insights

#### Foreign Currency Accounts Show Moderate Correlations Across Products
###### Highest relationships:
1. Foreign Currency ↔ Bank Deposits (0.41)
2. Foreign Currency ↔ Credit Card Balance (0.36)
3. Foreign Currency ↔ Bank Loans (0.36)

### Multicollinearity Observation

#### Potential Multicollinearity
###### Very high correlations:
1. Bank Deposits ↔ Checking Accounts (0.84)
2. Bank Deposits ↔ Saving Accounts (0.75)
